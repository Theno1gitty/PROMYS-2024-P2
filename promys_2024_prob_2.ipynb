```python
import random

# Set a random seed for reproducibility
#random.seed(42)

# Choose a random positive integer between 1 and 100
positive_integer = random.randint(1, 100)

# Choose a random negative integer such that its absolute value is smaller than positive_integer

if positive_integer > 1:
    negative_integer = random.randint(-positive_integer + 1, -1)
else:
    # Fallback if positive_integer is 1 (to satisfy the strict inequality condition)
    positive_integer = random.randint(2, 100)
    negative_integer = random.randint(-positive_integer + 1, -1)

print(f"Random positive integer: {positive_integer} (abs: {abs(positive_integer)})")
print(f"Random negative integer: {negative_integer} (abs: {abs(negative_integer)})")
```

```python
# Generate a series where term_(n+2) = term_n + term_(n+1)


series = [positive_integer, negative_integer]

# Generate the first 10 terms of the series
for _ in range(8):
    next_term = series[-2] + series[-1]
    series.append(next_term)

print("Sequence with the given starting integers:")
for idx, term in enumerate(series):
    print(f"term_{idx}: {term}")
```

```python
# Function to count the number of alternating signs at the beginning of the series
def count_alternating_signs(sequence):
    # Ignore 0
    filtered = [x for x in sequence if x != 0]

    if len(filtered) < 2:
        return 0

    alternation_count = 1
    for i in range(len(filtered) - 1):
        # Check if adjacent non-zero terms have opposite signs
        if filtered[i] * filtered[i+1] < 0:
            alternation_count += 1
        else:
            break

    return alternation_count

# Count the alternating signs for our series
alternating_count = count_alternating_signs(series)
print(f"Series: {series}")
print(f"Number of alternating signs at the start of the series (ignoring 0): {alternating_count}")
```
